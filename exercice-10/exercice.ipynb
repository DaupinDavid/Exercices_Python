```python
import pandas as pd

df = pd.read_csv("../data/titanic_propre.csv")


ports = pd.DataFrame({ "embark_town": ["Southampton", "Cherbourg", "Queenstown"], "pays": ["Angleterre", "France", "Irlande"], })

nb_avant = len(df)

df = df.merge(ports, on="embark_town", how="left", validate="many_to_one")

nb_avant == len(df)
df.head()
```

```python
print(f"Il y a {len(df)} passagers à bord du Titanic, dont {df['survived'].sum()} ont survécu.")

print("\n\n% de survie selon le pays d'embarquement :\n")
display(
    df.groupby('pays')
    .agg(
        passagers = ('pays', 'size'),
        taux_survie = ('survived', 'mean')
    )
    .assign(
        taux_survie = lambda x: (x['taux_survie'] * 100).round(2)
    )
)
```

Plus de la moitié des passagers provenant de France ont survécu contrairement à ceux provenant d'Angleterre ou d'Irlande avec un taux de survie aux alentours d'un tiers. 

```python
print("Table df : ")
display(
    df[['embark_town', 'pays']]
    .drop_duplicates()
    .sort_values('embark_town')
)

print("Table ports : ")
display(ports)

ports = ports[ports['embark_town'] != "Queenstown"]

print("Table ports après suppression de Queenstown : ")
display(ports)

df_inner = df.merge(ports, on="embark_town", how="inner", validate="many_to_one")
print(f"Il y a {len(df_inner)} passagers à bord du Titanic, dont {df_inner['survived'].sum()} ont survécu.")



```

On remarque que le INNER permet la fusion de 2 tables en conservant uniquement les lignes d'une valeur commune aux 2 tables de la colonne qui a permis la jointure.

```python
print("Table df avant la jointure :")
display(
    df[['embark_town', 'pays']]
    .drop_duplicates()
    .sort_values('embark_town')
)

print("Table ports :")
display(ports)

df_left = df.merge(
    ports,
    on="embark_town",
    how="left",
    validate="many_to_one"
)

print("Résultat de la df_left :")
display(
    df_left[['embark_town', 'pays_x', 'pays_y']]
    .drop_duplicates()
    .sort_values('embark_town')
)

print(f"Il y a {len(df_left)} passagers à bord du Titanic, dont {df_left['survived'].sum()} ont survécu.")
```

On remarque que le LEFT conserve toutes les valeurs de la table principale. Si une valeur est commune au 2 tables, les informations de cette valeur seront intégrées au df
