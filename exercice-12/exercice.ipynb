```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

df = sns.load_dataset("titanic")
os.makedirs("figures", exist_ok=True)

taxis = sns.load_dataset("taxis")
```

Quel âge avaient les passagers ?

Type de graphique : Histogramme (histplot)

Justification : On veut visualiser la distribution d'une variable quantitative (age).

Conclusion : La majorité des passagers avait entre 15 et 40 ans.

```python
fig, ax = plt.subplots(figsize=(10, 5))

sns.histplot(data=df, x="age", bins=30, ax=ax)

ax.set_title("La plupart des passagers avaient entre 15 et 40 ans")
ax.set_xlabel("Âge (années)")
ax.set_ylabel("Nombre de passagers")

fig.savefig("figures/1_ages_passagers.png", dpi=150, bbox_inches="tight")
plt.show()
```

La survie selon la classe et le sexe?

Type de graphique : Diagramme en barres (barplot) avec hue

Justification : On compare une mesure (taux de survie) selon des catégories (classe et sexe).

Conclusion : Les femmes présentent un taux de survie nettement supérieur à celui des hommes, quelle que soit la classe.

```python
fig, ax = plt.subplots(figsize=(10, 5))

sns.barplot(
    data=df,
    x="pclass",
    y="survived",
    hue="sex",
    ax=ax
)

ax.set_title("Les femmes ont davantage survécu dans toutes les classes")
ax.set_xlabel("Classe")
ax.set_ylabel("Taux de survie")

fig.savefig(
    "figures/2_survie_classe_sexe.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

Le prix du billet selon la classe?

Type de graphique : Boîte à moustaches (boxplot)

Justification : On compare la distribution d'une variable quantitative (fare) selon une catégorie (pclass).

Conclusion : Les passagers de 1ère classe ont payé des billets nettement plus chers que ceux des autres classes.

```python
fig, ax = plt.subplots(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="pclass",
    y="fare",
    ax=ax
)

ax.set_title("Les billets de 1ère classe coûtent beaucoup plus cher")
ax.set_xlabel("Classe")
ax.set_ylabel("Prix du billet ($)")

fig.savefig(
    "figures/3_prix_billet_classe.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

Taxis : le chiffre d'affaires jour après jour ?

Type de graphique : Courbe (lineplot)

Justification : On étudie une évolution au cours du temps.

Conclusion : Le chiffre d'affaires quotidien oscille généralement entre 3k et 5k $, avec quelques variations selon les jours.

```python
ca_jour = (
    taxis.groupby(taxis["pickup"].dt.date)["total"]
    .sum()
    .reset_index()
)

fig, ax = plt.subplots(figsize=(12, 5))

sns.lineplot(
    data=ca_jour,
    x="pickup",
    y="total",
    ax=ax
)

ax.set_title("Le chiffre d'affaires quotidien se situe majoritairement entre 3000 et 5000 $")
ax.set_xlabel("Date")
ax.set_ylabel("Chiffre d'affaires")

plt.xticks(rotation=45)

fig.savefig(
    "figures/4_chiffre_affaires_jour.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

Taxis : le lien entre distance et prix ?

Type de graphique : Nuage de points (scatterplot)

Justification : On cherche une relation entre deux variables quantitatives.

Conclusion : Plus la distance parcourue augmente, plus le prix total de la course tend à augmenter.

```python
fig, ax = plt.subplots(figsize=(10, 5))

sns.scatterplot(
    data=taxis,
    x="distance",
    y="total",
    ax=ax
)

ax.set_title("Les courses longues coûtent généralement plus cher")
ax.set_xlabel("Distance (miles)")
ax.set_ylabel("Prix total ($)")

fig.savefig(
    "figures/5_distance_prix.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```
