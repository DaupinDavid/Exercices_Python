```python
df.sort_values(by="fare", ascending=False)[["class", "embark_town", "fare"]].head(5)
```

Voici les 5 billets les plus chers avec la classe ainsi que le port d'embarquement.

```python
df[(df["sex"] == "female") & (df["class"] == "First")]
```

```python
femme_1ere_classe = (df[(df["sex"] == "female") & (df["class"] == "First")].shape[0])
passagers_1ere_classe = (df[df["class"] == "First"].shape[0])
pourcentage_femmes_1ere_classe = (femme_1ere_classe / passagers_1ere_classe) * 100

print(f"Pourcentage de femmes dans la première classe : {pourcentage_femmes_1ere_classe:.2f}%")
```

```python
df[(df["class"] == "First") & (df["sex"] == "female") & (df["alive"] == "yes")]
```

```python
femmes_1ere_classe_vivantes = (df[(df["class"] == "First") & (df["sex"] == "female") & (df["alive"] == "yes")].shape[0])

print(f"Nombre de femmes vivantes 1ère classe : {femmes_1ere_classe_vivantes}")
print(f"Nombre de femmes au total 1ère classe : {femme_1ere_classe}")
print(f"Pourcentage de femmes vivantes 1ère classe uniquement : {(femmes_1ere_classe_vivantes / femme_1ere_classe) * 100:.2f}%")
print(f"Pourcentage de femmes vivantes 1ère classe sur l'ensemble des passagers : {(femmes_1ere_classe_vivantes / df.shape[0]) * 100:.2f}%")
```

```python
hommes_3eme_classe = (df[(df["class"] == "Third") & (df["sex"] == "male")].shape[0])
hommes_3eme_classe_vivants = (df[(df["class"] == "Third") & (df["sex"] == "male") & (df["alive"] == "yes")].shape[0])

print(f"Nombre d'hommes vivants 3ème classe : {hommes_3eme_classe_vivants}")
print(f"Nombre d'hommes au total 3ème classe : {hommes_3eme_classe}")
print(f"Pourcentage d'hommes vivants 3ème classe uniquement : {(hommes_3eme_classe_vivants / hommes_3eme_classe) * 100:.2f}%") 
print(f"Pourcentage d'hommes vivants 3ème classe sur l'ensemble des passagers : {(hommes_3eme_classe_vivants / df.shape[0]) * 100:.2f}%")
```
